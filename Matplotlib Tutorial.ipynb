# Retail Sales Trend Analysis

```python
import matplotlib.pyplot as plt
```

```python
# Data loading

months = ["Jan", "Feb", "Mar", "Apr", "May", "Jun"]

sales = [120000, 145000, 138000, 165000, 190000, 215000]
```

```python
plt.plot(months, sales, color = 'blue', linestyle = '-', marker = 'o')
# create a line plot with blue color and a dashed line style

plt.xlabel('Month')
# create an x axis label

plt.ylabel('Sales (₦)')
# create a Y axis label

plt.title('Monthly Sales Trend')
# create a title on the chart

plt.show()
# shows the output chart
```

### Observation

####  It can be observed from the chart that there was a slow increase in sales from january to february followed by a dip in march before sales experienced a spontaneous increase in the remaining months.
#### The lowest sales occured in march and the highest occured in the month of Jume.


###  How it helps the Business

#### This visualization can help the owner identify how the sales trend for the first half went. This provides insights on the months that has the least sales and delve into the cause, consequently mitigating the probability of such re-ocurring in the second half of the business year 

# Customer Acquisition Comparison

```python
# Dataset

months = [1, 2, 3, 4, 5, 6]

campaign_a = [50, 65, 72, 90, 110, 125]

campaign_b = [45, 60, 68, 75, 95, 105]
```

```python
plt.plot(months, campaign_a, 'r-', marker = 's', label = 'Campaign A')
plt.plot(months, campaign_b, 'b-', marker = 'o', label = 'Campaign B')

plt.xlabel('Months in Numbers')
plt.ylabel('Customers Acquired')
plt.title('Customers Acquired by Month')
plt.xlim(1,6)
plt.legend()
plt.show()
```

###  Observation

#### Campaign A and B generated customers increasingly from the first months through 6 

#### According to Campaign A, after the promotion, the customers acquired increased significanly over the months expecially in the 4th month through 6 as compared to Campaign A, whose customer acquisition was almost similar during the first 3 months, however, it can be seen that although, they kept getting customers but compared to the former were less

```python
campaigns , axes = plt.subplots( nrows = 1, ncols = 2,  figsize = (13, 5))

# Campaign A chart

axes[0].plot(months,
             campaign_a,
             'r-',
             marker = 's')
axes[0].set_xlabel('Months in Numbers')
axes[0].set_ylabel('Customers Acquired')
axes[0].set_title('Customer Aquired through Campaign A')

# Campaign B chart

axes[1].plot(months,
              campaign_b,
              'b-',
              marker = 'o')
axes[1].set_xlabel ('Months in Numbers')
axes[1].set_ylabel('Customers Acquired')
axes[1].set_title('Customer Aquired through Campaign B')
plt.show()
```

#### Observation

#### According to Campaign A, after the promotion, the customers acquired increased significanly over the months expecially in the 4th month through 6 as compared to Campaign A, whose customer acquisition was almost similar during the first 3 months, however, it can be seen that although, they kept getting customers but compared to the former were less

# Healthcare Performance Dashboard

```python
# Dataset

months = [1, 2, 3, 4, 5, 6]

patients = [420, 450, 480, 510, 495, 540]

waiting_time = [65, 62, 58, 55, 60, 49]

satisfaction = [72, 74, 76, 78, 77, 83]

patient, axes = plt.subplots(nrows = 3, ncols = 1, figsize = (8, 12))

# first chart

axes[0].plot(months,
             patients,
             'r-',
             marker = 'o')
axes[0].set_xlabel('Months in Number')
axes[0].set_ylabel('Number of Patients')
axes[0].set_title('Number of Patients treated by Month')

# second chart

axes[1].plot(months,
            waiting_time,
            'g-',
            marker = 's')
axes[1].set_xlabel('Months in Number')
axes[1].set_ylabel('Average Waiting Time (mins)')
axes[1].set_title('Average Patient Waiting time by Month')

# Third chart

axes[2].plot(months,
            satisfaction,
            'b-',
            marker = '^')
axes[2].set_xlabel('Months in Number')
axes[2].set_ylabel('Customer Satisfaction %')
axes[2].set_title('Customer Satisfaction by Month')

patient.tight_layout()
plt.show()
```

#### Observation

#### The First chart showed a uniform increase in the number of patients treated for the first 4 months, a trend which suffered a dip from the 4th to 5th month followed by an increase in the the last month recorded

#### The Second chart showed a decline the average patient waiting time  from the first month through the fourth month, this which later went through an incline through the 4th to 5th month then dipped

#### The third chart showed a uniform percentage increase from the first month through the fourth and then suffered a decline in the 4th to 5th months before increasing considerably again in the last month

#### Combining the charts, it was clear that, although the number of patients dropped in the 4th and 5th month (510-495), the average waiting time increased (55-60 mins), consequently reducing the customer satisfaction (78-77 %)

#### It is however important to note that, the decline only occured once in the chart and is not enough sample to speak to the popoulation of patients. I recommend that we collect more data across various time period to further gain more insights into customer bahaviour.

# Project title: Visualizing Business Performance with Matplotlib

```python
# Project Dataset

months = [1, 2, 3, 4, 5, 6]

revenue = [150000, 165000, 158000, 190000, 215000, 240000]

customers = [80, 92, 88, 110, 125, 140]

product_a = [45, 50, 48, 60, 68, 75]

product_b = [35, 40, 42, 45, 52, 58]

sales, axes = plt.subplots(nrows = 3, ncols = 1, figsize = (8, 12))

# chart A

axes[0].plot(months,
             revenue,
             'c-',
             marker = 'x')
axes[0].set_xlabel('Months in Number')
axes[0].set_ylabel('Revenue (₦)')
axes[0].set_title('Revenue Generated by Month')

# Chart B

axes[1].plot(months,
             customers,
             'k-',
             marker = 's')
axes[1].set_xlabel('Months in Number')
axes[1].set_ylabel('Number of Customers')
axes[1].set_title('Customer Growth by Month')


# Chart C
axes[2].plot(months, product_a, 'b-', marker = 's', label = 'Product A')
axes[2].plot(months, product_b, 'r-', marker = 'o', label = 'Product B')
axes[2].set_xlabel('Months in Number')
axes[2].set_ylabel('Product Quantity')
axes[2].set_title('Performance of Product A vs Product B')
plt.legend()

sales.tight_layout()
```

#### Observation

#### 1. What is the overall pattern in revenue?
##### The revenue went through a slight increment then dipped before going through a uniform increase for the remainder of the month

#### 2. Which months show a decline compared with the preceding month?
##### The third month experienced a decline as compared to subsequent months

#### 3. How does the customer trend compare with the revenue trend?
##### The customer trend has a similar trend with the revenue trend and this could imply that there is a direct relation ship between those two attributes, meaning, an increase in the number of customers could also imply a gains in terms of revenue for the company

#### 4. Which product records higher monthly sales in the final month?
##### According to the comparison chart above, it can be deduced that the products which recorded the higher monthly sales was product A with the quantity sold being 75 units

#### 5. Are there months when revenue and customer numbers move in different directions?
##### It is also visible from the chart that the trend of revenue amd customer counts moves in a similar pattern , hence, at no point did any deviate from each other.

#### 6. What additional information would help the business owner understand the observed patterns?
##### According to the charts, there might be need for data collection over more periods of time to determine why customers behaviour towards the products and see if interventions can be made to increase customers interest in product B as well
